```python
import pandas as pd

submission = pd.read_csv('notebooks/solution.csv')
test_features = pd.read_csv('data/test_set.csv')

merged = test_features.merge(submission[['Pubchem_ID', 'taste_cluster']], on='Pubchem_ID', how='left')

merged['taste_cluster'] = merged['taste_cluster'].fillna(0).astype(int)

merged[['taste_cluster']].to_csv('submission.csv', index=False)

print("Файл submission.csv успешно создан.")
```
