```python
import matplotlib.pyplot as plt
import numpy as np

def visualize_class_images(train_df, images, n_per_class=10):
    """
    Для каждого класса визуализирует n_per_class молекул.

    Args:
        train_df: pd.DataFrame с колонками "SMILES" и "target"
        images: numpy array формы (N, H, W, 3)
        n_per_class: сколько изображений показывать на класс
    """
    classes = train_df['taste_cluster'].unique()
    
    for cls in classes:
        idxs = np.where(train_df['taste_cluster'] == cls)[0]
        selected_idxs = idxs[:n_per_class]  # первые n_per_class
        n_cols = min(n_per_class, len(selected_idxs))
        
        plt.figure(figsize=(3 * n_cols, 3))
        plt.suptitle(f"Class {cls}", fontsize=16)
        
        for i, idx in enumerate(selected_idxs):
            plt.subplot(1, n_cols, i + 1)
            img = images[idx]
            if img.shape[-1] == 1:
                img = img.squeeze(-1)
                plt.imshow(img, cmap='gray')
            else:
                plt.imshow(img)
            plt.axis('off')
            plt.title(train_df['SMILES'].iloc[idx], fontsize=8)
        
        plt.tight_layout()
        plt.show()


# Пример использования
train_df = pipeline.get_processed_data()[0]
train_images = pipeline.get_images()[0]  # (N, 244, 244, 3)
visualize_class_images(train_df, train_images, n_per_class=10)
```
